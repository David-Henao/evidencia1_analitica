```python
#  Traigo las clases que armamos en los otros archivos sueltos
from traer_datos import CargadorHF
from hacer_resumen import ResumidorDataset
```

```python
# Cargo el dataset Iris de Hugging Face usando nuestra clase
conector = CargadorHF(dataset_id='scikit-learn/iris', split_name='train')

# Guardamos los datos originales aca
df_original = conector.obtener_dataframe()

# Mostramos las primeras filas para verificar que todo cargo bien
df_original.head()
```

```python
# Cambio los nombres de las columnas a español para que se entienda mejor
diccionario_nombres = {
    'Id': 'id',
    'SepalLengthCm': 'largo_sepalo',
    'SepalWidthCm': 'ancho_sepalo',
    'PetalLengthCm': 'largo_petalo',
    'PetalWidthCm': 'ancho_petalo',
    'Species': 'especie'
}

# Aplico el renombrado en un dataframe nuevo
df_transformado = df_original.rename(columns=diccionario_nombres)

# Muestro los primeros datos modificados
df_transformado.head()
```

```python
# Llamo al resumidor y le paso la tabla con los nombres nuevos
analizador = ResumidorDataset(df_transformado)

# Le digo que cree el archivo de texto final con el resumen
analizador.guardar_resumen_txt("resumen_dataset.txt")
```
